## ExactLinearODE Example

This notebook demonstrates the package-level interface for reproducing the Exact Linear ODE benchmark runs and reporting network-recovery AUROC/AUPR metrics.


```python
import os
import torch
```

```python

from exact_linear_ode import ExactLinearODEBenchmark
```

We import `ExactLinearODEBenchmark`, the package-level class for listing configured datasets and running benchmark experiments.


Available datasets and solvers are exposed on the class, so you can inspect them without creating a benchmark object.

Supported methods include PHOENIX (`phoenix`), which uses the vendored no-prior neural ODE implementation. The dataset configs select smoothed expression inputs and the same defaults as the five PHOENIX reproduction notebooks.


```python
print('datasets', ExactLinearODEBenchmark.datasets)
print('solvers', ExactLinearODEBenchmark.solvers)
```

Run every exposed solver across every configured dataset. The ExactLinearODE solver follows each dataset config: hyperparameter selection, final refit on the full ordered dataset, reference-network evaluation, and perturbation AUROC for `ours`.


```python
os.environ["PATH"] = "/N/soft/rhel8/r/gnu/4.5.1/bin:" + os.environ["PATH"]
```

```python
solvers = ['ours', 'genie3', 'dyn_genie3', 'scode', 'phoenix', 'grisli', 'prob', 'swing', 'lasso', 'mi', 'lm', 'ida_min', 'ida_mean']
results = ExactLinearODEBenchmark(dataset='data3').run(solvers=solvers, progress=True)
print(results[['solver', 'status', 'message', 'auroc', 'auroc_absA', 'auroc_perturb']].to_string(index=False))
```

The same interface can run several configured datasets. Full runs can take a long time, so this cell is left as an optional example.


```python
# The full all-dataset run is executed above.
results
```

To run only one method, pass an explicit solver list.
For PHOENIX, use `ExactLinearODEBenchmark(dataset="data3").run(solvers=["phoenix"])`.
