### Fitted expression trajectories

```python
from diode import DIODEBenchmark
from diode.utils.plotting import plot_expression_trajectories

datasets = ["data4_trophectoderm"]
solvers = ["DIODE", "SCODE","PHOENIX"]
genes = ["ZFAT", "GATA3", "KLF10", "ZNF18", "DUSP6"]

benchmark = DIODEBenchmark()
metrics = benchmark.run(datasets=datasets, solvers=solvers, skip_unavailable=False, progress=True)
```

```python
predictions = benchmark.predict(genes=genes)
figures = plot_expression_trajectories(predictions)
```
