## Reproduce Fig.2f — Mouse MEF-iN

This notebook runs the DIODE benchmark on data2 (Mouse MEF-iN) and reports solver AUROC results.


```python
import os
import torch
```

```python
from diode import DIODEBenchmark
```

We import `DIODEBenchmark`, the package-level class for listing configured datasets and running benchmark experiments.


Available datasets and solvers are exposed on the class, so you can inspect them without creating a benchmark object.


```python
print('datasets', DIODEBenchmark.datasets)
print('solvers', DIODEBenchmark.solvers)
```

Run the selected solvers on data2. DIODE follows the dataset config: hyperparameter selection, final refit on the full ordered dataset, reference-network evaluation, and AUROC.


```python
solvers = ['DIODE', 'GENIE3', 'GRNBoost2', 'dynGENIE3', 'SCODE', 'PHOENIX', 'GRISLI', 'PROB', 'SWING', 'Lasso', 'MI', 'LM', 'IDA_min', 'IDA_mean']
results = DIODEBenchmark(dataset='data2').run(solvers=solvers, progress=True)
display_results = results[['solver', 'auroc']].rename(columns={'auroc': 'AUROC'})
diode = results['solver'] == 'DIODE'
display_results.loc[diode, 'AUROC'] = results.loc[diode, 'auroc_perturb']
print(display_results.to_string(index=False))
```

The same interface can run several configured datasets. Full runs can take a long time, so this cell is left as an optional example.
