# Data4 Epiblast

Clean replication notebook for the DIODE workflow.

```python
import pandas as pd

from diode import DIODEBenchmark
from diode.utils.io import load_config, resolve_path

benchmark = DIODEBenchmark(dataset='data4_epiblast')
CONFIG_PATH = benchmark.config_path('data4_epiblast')
REPO_ROOT = benchmark.repo_root
config = load_config(CONFIG_PATH)
config
```

## Run Dataset

This cell runs the dataset using the committed config. For a quick smoke test, temporarily lower `training.epochs` in the config or use an external temporary config.

```python
metrics = benchmark.run(solvers=['DIODE'])
metrics[['auroc_perturb']].rename(columns={'auroc_perturb': 'auroc'})
```
