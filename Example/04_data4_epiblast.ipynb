# Epiblast Trajectory (Data4 Epiblast)

```python
from diode import DIODEBenchmark
```

## Run Dataset


```python
solvers = ['DIODE']
# dataset = data4_Epiblast is Epiblast Trajectory
results = DIODEBenchmark(dataset='data4_epiblast').run(solvers=solvers, progress=True)
display_results = results[['solver', 'auroc']].rename(columns={'auroc': 'AUROC'})
diode = results['solver'] == 'DIODE'
display_results.loc[diode, 'AUROC'] = results.loc[diode, 'auroc_perturb']
display_results
```
