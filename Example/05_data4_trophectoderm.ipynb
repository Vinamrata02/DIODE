# Trophectoderm Trajectory (Data4 - Trophectoderm)

```python
from diode import DIODEBenchmark
```

## Run Dataset


```python
solvers = ['DIODE']
# dataset = data4_trophectoderm is Trophectoderm Trajectory
results = DIODEBenchmark(dataset='data4_trophectoderm').run(solvers=solvers, progress=True)
display_results = results[['solver', 'auroc']].rename(columns={'auroc': 'AUROC'})
diode = results['solver'] == 'DIODE'
display_results.loc[diode, 'AUROC'] = results.loc[diode, 'auroc_perturb']
display_results
```
